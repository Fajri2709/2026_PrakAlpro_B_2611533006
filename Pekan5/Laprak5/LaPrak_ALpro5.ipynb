## 2.1 Dasar Teori

### 2.1.1 Pengertian Python
Python merupakan bahasa pemrograman tingkat tinggi yang bersifat general-purpose. Python dirancang dengan sintaks yang sederhana sehingga mudah dipahami dan digunakan. Python dapat digunakan untuk berbagai kebutuhan, seperti pengembangan aplikasi, otomatisasi, pengolahan data, kecerdasan buatan, dan pembelajaran pemrograman.

Salah satu keunggulan Python adalah struktur sintaksnya yang relatif sederhana. Python juga menggunakan indentasi untuk menentukan struktur blok kode, sehingga program dapat terlihat lebih rapi dan mudah dibaca.

### 2.1.2 Pengertian Perulangan
Perulangan atau looping adalah proses menjalankan suatu blok kode secara berulang selama kondisi atau aturan tertentu terpenuhi.

Sebagai contoh, apabila program ingin menampilkan angka 1 sampai 5, programmer dapat menuliskan perintah print() sebanyak lima kali. Namun, cara tersebut tidak efektif. Dengan menggunakan perulangan, programmer cukup menuliskan satu perintah yang kemudian dijalankan beberapa kali.

Python memiliki dua jenis perulangan utama, yaitu:

1. Perulangan for

2. Perulangan while

Pada praktikum ini pembahasan difokuskan pada perulangan for.

### 2.1.3 Pengertian Perulangan For
Perulangan for digunakan untuk melakukan iterasi terhadap sebuah objek yang dapat diiterasi (iterable). Pada setiap proses iterasi, variabel perulangan akan menyimpan nilai dari elemen yang sedang diproses.

Bentuk umum perulangan for adalah:

for variabel in iterable:
    perintah

Contoh:

for i in range(5):

    print(i)

Output:

0

1

2

3

4

Pada program tersebut, variabel i akan mengambil nilai dari setiap elemen yang dihasilkan oleh range(5).

### 2.1.4 Fungsi range()
Fungsi range() digunakan untuk menghasilkan urutan bilangan dan sering digunakan bersama perulangan for.

Bentuk dasar fungsi range() adalah:

range(stop)

Selain itu, range() juga dapat menggunakan nilai awal dan langkah:

range(start, stop)

dan:

range(start, stop, step)

Contoh:

for angka in range(1, 6):

    print(angka)

Output:

1

2

3

4

5

Angka terakhir pada range() tidak termasuk dalam hasil. Dengan demikian, range(1, 6) menghasilkan angka mulai dari 1 sampai 5.

Contoh penggunaan step:

for angka in range(0, 11, 2):

    print(angka)

Output:

0

2

4

6

8

10

Pada contoh tersebut, nilai bertambah dua pada setiap iterasi.

### 2.1.5 Perulangan For pada List
Perulangan for dapat digunakan untuk mengakses setiap elemen dalam sebuah list.

Contoh:

nama = ["Andi", "Budi", "Citra", "Dina"]

for orang in nama:

    print(orang)

Output:

Andi

Budi

Citra

Dina

Pada program tersebut, variabel orang akan mengambil setiap nilai yang terdapat dalam list nama.

### 2.1.6 Perulangan For pada String
String juga termasuk objek yang dapat diiterasi. Oleh karena itu, setiap karakter dalam string dapat diproses menggunakan perulangan for.

Contoh:

kata = "PYTHON"

for huruf in kata:

    print(huruf)

Output:

P

Y

T

H

O

N

Program tersebut membaca setiap karakter dalam string PYTHON kemudian menampilkannya satu per satu.

### 2.1.7 Perulangan For dengan Percabangan If
Perulangan for dapat digabungkan dengan percabangan if. Kombinasi tersebut berguna ketika hanya elemen tertentu yang ingin diproses.

Contoh:

for angka in range(1, 11):

    if angka % 2 == 0:

        print(angka)

Output:

2

4

6

8

10

Program tersebut menampilkan bilangan genap dari 1 sampai 10. Kondisi angka % 2 == 0 digunakan untuk memeriksa apakah suatu angka habis dibagi dua.


### 2.1.8 Nested For
Nested for adalah perulangan for yang berada di dalam perulangan for lainnya. Konsep ini biasanya digunakan ketika program perlu mengolah data dalam dua tingkat atau membuat pola tertentu.

Contoh:

for i in range(1, 4):

    for j in range(1, 4):

        print(i, j)

Output:

1 1

1 2

1 3

2 1

2 2

2 3

3 1

3 2

3 3

Pada program tersebut, perulangan bagian dalam dijalankan sampai selesai untuk setiap satu kali proses pada perulangan bagian luar.

### 2.2.3 Program "perulangan_for3_2611533006"

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan format input

ulang_3006 = int(input("Masukkan jumlah perulangan: "))

jumlah_3006 = 0
for i_3006 in range(1, ulang_3006 + 1):
    print(i_3006, end=" ")
    jumlah_3006 = jumlah_3006 + i_3006

    if i_3006 < ulang_3006:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_3006, end="")
print()
print("Jumlah =", jumlah_3006)
```

**Penjelasan :**

2. **Baris 2:** Membuat variabel `jumlah_3006` dengan nilai awal `0` sebagai tempat menyimpan hasil penjumlahan.
3. **Baris 3:** Melakukan perulangan dari angka `1` sampai `ulang_3006`.
4. **Baris 4:** Menampilkan angka perulangan dalam satu baris menggunakan `end=" "`.
5. **Baris 5:** Menambahkan nilai `i_3006` ke dalam `jumlah_3006`.
6. **Baris 6:** Mengecek apakah nilai `i_3006` masih kurang dari jumlah perulangan.
7. **Baris 7:** Jika masih ada angka berikutnya, program menampilkan tanda `" + "`.
8. **Baris 8:** Jika sudah mencapai angka terakhir, program menampilkan tanda `" = "` dan hasil penjumlahan `jumlah_3006`.
9. **Baris 9:** Membuat baris baru setelah proses perulangan selesai.
10. **Baris 10:** Menampilkan hasil akhir penjumlahan dengan format `Jumlah = ...`.

### 2.2.5 Program "nested_for1_2611533006"

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan format input

batas_3006 = int(input("Masukkan nilai batas: "))
for line_3006 in range(1, batas_3006 + 1):
    for j_3006 in range(1, (-1 * line_3006 + batas_3006) + 1):
        print(".", end="")
    print(line_3006)
```

### 2.2.8 Program "nested_for4_2611533006"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan format input

tinggi_3006 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3006 %2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3006 = tinggi_3006
    c_3006 = a_3006
    lebar_3006 = (2 * tinggi_3006) - 2

    for i_3006 in range(1, tinggi_3006 + 1):
        b_3006 = c_3006 + 1

        for j_3006 in range(1, lebar_3006 + 1):

            # Baris atas dan bawah
            if i_3006 == 1 or i_3006 == tinggi_3006:
                if j_3006 == 1 or j_3006 == lebar_3006:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris Isi
            else:
                if j_3006 == 1 or j_3006 == lebar_3006:
                    print("|", end="")
                else:
                    if j_3006 == c_3006:
                        print("<", end="")
                    elif j_3006 == b_3006:
                        print(">", end="")
                    elif j_3006 == (lebar_3006 - c_3006):
                        print("<", end="")
                    elif j_3006 == (lebar_3006 - c_3006 + 1):
                        print(">", end="")
                    elif j_3006 > b_3006 and j_3006 < (lebar_3006 - c_3006):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_3006 -= 2

        if a_3006 <= 0:
            c_3006 = (-a_3006) + 2
        else:
            c_3006 = a_3006
```

Berdasarkan pembahasan yang telah dilakukan, dapat disimpulkan bahwa perulangan for merupakan salah satu struktur dasar dalam bahasa pemrograman Python yang digunakan untuk menjalankan suatu blok kode secara berulang terhadap objek yang dapat diiterasi.

Perulangan for dapat digunakan bersama fungsi range() untuk melakukan perulangan berdasarkan urutan bilangan. Selain itu, perulangan for dapat digunakan untuk mengakses elemen pada list, tuple, string, dan objek iterable lainnya.
